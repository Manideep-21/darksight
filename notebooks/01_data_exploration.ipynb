```python
import sys; sys.path.insert(0, '..')
import pandas as pd, matplotlib.pyplot as plt
meta = pd.read_csv('../data/exdark_meta.csv', dtype={'stem': str})
print(open('../reports/data_stats.md').read())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
meta.groupby(['light_name', 'split']).size().unstack().plot.bar(ax=axes[0], title='images per lighting condition')
meta.n_boxes.clip(upper=20).plot.hist(bins=20, ax=axes[1], title='boxes per image (clipped at 20)')
(meta.w * meta.h / 1e6).plot.hist(bins=30, ax=axes[2], title='megapixels after resize')
plt.tight_layout()
```

```python
# brightness distribution: how dark is 'low-light' really? (informs the gate tau)
from darksight.io_utils import load_rgb
from darksight.enhance import mean_luminance
test = meta[meta.split == 'val'].sample(300, random_state=0)
test['lum'] = [mean_luminance(load_rgb(f'../data/exdark_yolo/raw/images/val/{s}.png')) for s in test.stem]
test.boxplot(column='lum', by='light_name', rot=45, figsize=(10, 4)); plt.ylabel('mean luminance')
```
